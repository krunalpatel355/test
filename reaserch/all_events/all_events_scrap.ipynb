```python
import re
import csv
import time
import requests
import threading
import pandas as pd
from dateutil import parser
from bs4 import BeautifulSoup
from concurrent.futures import ThreadPoolExecutor, as_completed
```

```python
# List of URLs to scrape
urls = {
    'united states': 'https://allevents.in/location.php?country=united%20states',
    'canada': 'https://allevents.in/location.php?country=canada'
}

# User-Agent headers to simulate browser
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'
}

# List to collect all links
all_links = []

# Scrape each URL
for country, url in urls.items():
    print(f"Scraping {country}...")
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, 'html.parser')

    container = soup.find('div', class_='loc-container')
    if container:
        list_items = container.find_all('li')
        for li in list_items:
            a_tag = li.find('a', href=True)
            if a_tag:
                all_links.append({
                    'Country': country,
                    'Link': a_tag['href']
                })

# Save all links to CSV
with open('states_data_links.csv', 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=['Country', 'Link'])
    writer.writeheader()
    writer.writerows(all_links)

print("Links saved to states_data_links.csv")
```

```python
input_csv = 'states_data_links.csv'
output_csv = 'events_data_links.csv'
max_workers = 10  # Adjust threads as needed
lock = threading.Lock()

HEADERS = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko)'
                  ' Chrome/114.0.0.0 Safari/537.36'
}

def scrape_events(country, url):
    events = []
    try:
        response = requests.get(url, headers=HEADERS)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, 'html.parser')
        event_sections = soup.find_all('div', class_='event-fold')

        for section in event_sections:
            # Event Type heading, ignoring <span>
            event_type_tag = section.find('h2', class_='headstyle-1 acolor')
            if event_type_tag:
                event_type = ''.join(event_type_tag.find_all(text=True, recursive=False)).strip()
            else:
                event_type = 'Unknown Event Type'

            # Sometimes event cards may be under different class or tag
            # Try multiple approaches to find event cards
            event_cards = section.find_all('li', class_='event-card event-card-link')
            if not event_cards:
                # fallback to finding any 'li' inside section
                event_cards = section.find_all('li')

            if not event_cards:
                continue  # no events under this type, skip

            for card in event_cards:
                # Try to get title div and link robustly
                title_div = card.find('div', class_='title')
                if not title_div:
                    title_div = card.find('a', class_='event-card-link')  # fallback

                title = title_div.get_text(strip=True) if title_div else 'No Title'

                link_tag = title_div.find('a') if title_div else card.find('a')
                link = link_tag['href'] if link_tag and link_tag.has_attr('href') else 'No Link'

                events.append([country, event_type, title, link])

        # Check for pagination (if exists)
        next_page = soup.find('a', class_='next')
        if next_page and next_page.has_attr('href'):
            next_url = next_page['href']
            if next_url.startswith('/'):
                # Build full URL from relative path
                base_url = '/'.join(url.split('/')[:3])
                next_url = base_url + next_url
            # Add small delay to be polite
            time.sleep(1)
            events.extend(scrape_events(country, next_url))

    except Exception as e:
        print(f"Failed to scrape {url}: {e}")

    return events


def main():
    with open(input_csv, newline='', encoding='utf-8') as infile:
        reader = list(csv.DictReader(infile))

    with open(output_csv, 'w', newline='', encoding='utf-8') as outfile:
        writer = csv.writer(outfile)
        writer.writerow(['country', 'event_type', 'title', 'link'])

    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = []
        for row in reader:
            country = row['Country']
            url = row['Link']
            futures.append(executor.submit(scrape_events, country, url))

        with open(output_csv, 'a', newline='', encoding='utf-8') as outfile:
            writer = csv.writer(outfile)
            for future in as_completed(futures):
                events = future.result()
                if events:
                    with lock:
                        writer.writerows(events)

    print(f"Scraping complete. Data saved to '{output_csv}'")


if __name__ == '__main__':
    main()
```

```python
INPUT_CSV = 'events_data_links.csv'
OUTPUT_CSV = 'events.csv'

def safe_text(soup, selector):
    tag = soup.select_one(selector)
    return tag.get_text(strip=True) if tag else ''

def find_price(soup):
    possible_texts = []

    # 1. Check common tags that may contain price info
    tags = soup.find_all(['p', 'span', 'div'])
    for tag in tags:
        text = tag.get_text(strip=True).lower()
        if any(keyword in text for keyword in ['cad', 'free', 'n/a', 'no cost', '$']):
            possible_texts.append(text)

    # 2. Try to find relevant price value
    for text in possible_texts:
        if 'starting at cad' in text:
            return text.replace('starting at cad', '').replace('cad', '').strip().upper()
        if 'cad' in text:
            return text.replace('cad', '').strip().upper()
        if 'free' in text:
            return 'FREE'
        if 'no cost' in text:
            return 'FREE'
        if 'n/a' in text:
            return 'N/A'
        if '$' in text:
            return text.strip()

    return 'N/A'  # Default if no recognizable price is found

def find_image(soup):
    img_tag = soup.find('img', class_='event-banner')
    return img_tag['src'] if img_tag and img_tag.has_attr('src') else ''

def scrape_event_details(row):
    url = row['link']
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
    except Exception as e:
        print(f"[ERROR] Failed to fetch {url}: {e}")
        return {**row, 'image_url': '', 'host': '', 'full_date_time': '', 'event_place': '', 'full_address': '', 'ticket_price': 'N/A'}

    soup = BeautifulSoup(response.text, 'html.parser')

    return {
        **row,
        'image_url': find_image(soup),
        'host': safe_text(soup, 'h2.eps-org-name'),
        'full_date_time': safe_text(soup, 'p.event-time-label'),
        'event_place': safe_text(soup, 'p.event-location'),
        'full_address': safe_text(soup, 'p.event-location + p'),
        'ticket_price': find_price(soup)
    }

def main():
    df = pd.read_csv(INPUT_CSV)

    results = []

    with ThreadPoolExecutor(max_workers=10) as executor:
        future_to_row = {executor.submit(scrape_event_details, row): row for _, row in df.iterrows()}
        for future in as_completed(future_to_row):
            row = future_to_row[future]
            try:
                data = future.result()
                results.append(data)
                print(f"Scraped: {data['title']} - {data['ticket_price']}")
            except Exception as e:
                print(f"[ERROR] Exception for {row['link']}: {e}")

    final_df = pd.DataFrame(results)
    final_df.to_csv(OUTPUT_CSV, index=False)
    print(f"✅ Saved detailed event data to {OUTPUT_CSV}")

if __name__ == '__main__':
    main()
```

```python
# Load the original CSV
df = pd.read_csv('events.csv')

# Function to parse full_date_time into start_date, end_date, start_time, end_time
def parse_datetime(row):
    full_date_time = str(row['full_date_time'])

    # Try to split time range from date string
    if ' to ' in full_date_time:
        parts = full_date_time.split(' to ')
        start_raw = parts[0]
        end_raw = parts[1].split('(')[0].strip()  # Remove timezone
    elif ' - ' in full_date_time:
        parts = full_date_time.split(' - ')
        start_raw = parts[0]
        end_raw = parts[1].split('(')[0].strip()
    else:
        start_raw = full_date_time.split('(')[0].strip()
        end_raw = start_raw

    try:
        start_dt = parser.parse(start_raw, fuzzy=True)
        end_dt = parser.parse(end_raw, fuzzy=True)
    except:
        return pd.Series(['', '', '', ''])

    return pd.Series([
        start_dt.strftime('%Y-%m-%d'),
        end_dt.strftime('%Y-%m-%d'),
        start_dt.strftime('%H:%M'),
        end_dt.strftime('%H:%M')
    ])

# Function to extract state and postal code
def extract_address_parts(address):
    address = str(address)
    state_match = re.search(r'\b[A-Z]{2}\b', address)
    postal_match = re.search(r'\b\d{5}(-\d{4})?\b', address)

    state = state_match.group() if state_match else ''
    postal_code = postal_match.group() if postal_match else ''

    return pd.Series([postal_code, state])

# Parse datetime into new columns
df[['start_date', 'end_date', 'start_time', 'end_time']] = df.apply(parse_datetime, axis=1)

# Extract postal_code and state_name
df[['postal_code', 'state_name']] = df['full_address'].apply(extract_address_parts)

# Create final DataFrame with required columns and rename
df_final = pd.DataFrame({
    'event_title': df['title'],
    'summary': '',  # summary not available
    'image_url': df['image_url'],
    'language': 'English',  # Assuming all events are in English
    'event_type': df['event_type'],
    'event_host': df['host'],
    'ticket_price': df['ticket_price'],
    'booking_url': df['link'],
    'start_date': df['start_date'],
    'end_date': df['end_date'],
    'start_time': df['start_time'],
    'end_time': df['end_time'],
    'event_place': df['event_place'],
    'full_address': df['full_address'],
    'country_name': df['country'],
    'state_name': df['state_name'],
    'city_name': '',  # Not provided in structured way
    'postal_code': df['postal_code']
})

# Save to new CSV
df_final.to_csv('events_cleaned.csv', index=False)
```
